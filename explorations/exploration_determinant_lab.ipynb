## Experiment 4. When is a matrix *not* invertible?

This is **Example 3.16**: for which values of $\lambda$ is $C = \begin{bmatrix} 5 & 0 & 0 \\ 0 & 0 & 1 \\ 1 & 1 & 0 \end{bmatrix} - \lambda I_3$ not invertible?

Instead of solving by hand, let's scan through values of $\lambda$ and ask panchi two questions for each: what's the determinant, and is the matrix invertible? (`lam` is short for $\lambda$, and `pan.identity(3)` is $I_3$.)

```python
M = pan.Matrix([
    [5, 0, 0],
    [0, 0, 1],
    [1, 1, 0],
])

for lam in range(-3, 8):
    C = M - lam * pan.identity(3)
    invertible = pan.is_invertible(C)
    print(f"lambda = {lam:3}   det(C) = {det(C):5}   invertible? {invertible}")
```

Here is the same idea as a picture: $\det(C)$ for *every* $\lambda$ between $-3$ and $7$. The marked points are where the curve crosses zero. (The code that draws it is hidden; double-click the cell if you're curious.)

```python
# @title Plot det(M - lambda I)
import matplotlib.pyplot as plt

lams = [x / 20 for x in range(-60, 141)]
dets = [det(M - lam * pan.identity(3)) for lam in lams]
roots = [lam for lam in range(-3, 8) if det(M - lam * pan.identity(3)) == 0]

fig, ax = plt.subplots(figsize=(6, 3))
ax.axhline(0, color="#8a8983", linewidth=1)
ax.plot(lams, dets, color="#2a78d6", linewidth=2)
ax.scatter(roots, [0] * len(roots), s=64, color="#2a78d6",
           edgecolor="white", linewidth=2, zorder=3)
for r in roots:
    ax.annotate(rf"$\lambda = {r}$", (r, 0), textcoords="offset points", xytext=(0, 10),
                ha="center", color="#52514e")
ax.set_xlabel(r"$\lambda$", color="#52514e")
ax.set_ylabel(r"$\det(M - \lambda I)$", color="#52514e")
ax.set_title(r"$\det(M - \lambda I)$ as $\lambda$ changes", color="#0b0b0b", loc="left")
ax.set_ylim(-40, 40)
ax.grid(color="#e4e3df", linewidth=0.8)
ax.tick_params(colors="#52514e")
for side in ["top", "right"]:
    ax.spines[side].set_visible(False)
plt.show()
```
